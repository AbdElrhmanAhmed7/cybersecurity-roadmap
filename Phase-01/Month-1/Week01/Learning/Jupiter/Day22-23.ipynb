## Day 22 Regex


```python
import re
```

```python
text = "Catch me if you can. there is a Caterpillar"
pattern = re.compile(r"Cat(erpillar|astrophe|ch|egory)")
pattern.findall(text)
```

```python
text = "Cell: 415-555-9999 Work: 212-555-0000"
pattern = re.compile(r'\d{3}-\d{3}-\d{4}')
pattern.findall(text)
```

```python
text = "Cell: 415-555-9999 Work: 212-555-0000"
pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
pattern.findall(text)
```

```python
>>> pattern = re.compile(r'\d{3}')
>>> pattern.findall('123456798')
```

```python
for x in pattern.finditer("123456798"):
    print(x)
```

# Day 23 (regex continue)


```python
vowel_pattern = re.compile(r"[^a-zA-z]")
vowel_pattern.findall("RoboCop eats BABY FOOD.")
```

```python
>>> pattern = re.compile(r'\d+\s\w+')
>>> pattern.findall('12 drummers, 11 pipers, 10 lords, 9 ladies, 8 maids, 7 swans, 6 geese, 5 rings, 4 birds, 3 hens, 2 doves, 1 partridge')
```

```python
>>> at_re = re.compile(r'.at')
>>> at_re.findall('The cat in the hat sat on the flat mat.')
```

```python
>>> pattern = re.compile(r'42?!')
>>> pattern.search('4!')
```

```python
>>> pattern = re.compile(r'(\d{3}-)?\d{3}-\d{4}')
>>> match1 = pattern.search('My number is 555-4242')
>>> match1.group()
```

```python
>>> pattern = re.compile('Eggs(and spam)*')
>>> pattern.search('Eggs')
```

```python
pattern = re.compile(r'Eggs( and spam)+')
pattern.search('Eggs and spam and spam')
```

```python
pattern = re.compile(r"(Ha){3,5}")
pattern.search("Ha" * 4)
```

```python
pattern = re.compile(r"(Ha){3,5}?")
pattern.search("Ha" * 5)
```

```python
>>> lazy_pattern = re.compile(r'<.*?>')
>>> match1 = lazy_pattern.search('<To serve man> for dinner.>')
>>> match1.group()
```

```python
>>> greedy_re = re.compile(r'<.*>')
>>> match2 = greedy_re.search('<To serve man> for dinner.>')
>>> match2.group()
```

```python
>>> pattern = re.compile(r'\bcat.*?\b')
>>> pattern.findall('The cat found a catapult catalog in the catacombs.')
```

```python
re.search(r"(\w)(\w)(\w)(\w)(\w)--\5\4\3\2\1", "Alien--neilA")
```
