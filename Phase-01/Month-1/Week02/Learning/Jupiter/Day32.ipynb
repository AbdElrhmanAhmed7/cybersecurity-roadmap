# Day 32 (subprocess module)


```python
import subprocess
```

```python

process = subprocess.Popen(
    ["python", "--version"],
    stdout=subprocess.PIPE,
    text=True
)

process.communicate()
```

```python
try:
    output = subprocess.check_output(["python", "--version"], text=True)
    print(output)
except subprocess.CalledProcessError:
    print("Command failed")
```

```python
code = subprocess.call(["python", "--version"])

if code == 0:
    print("Success")
else:
    print("Failed")
```

```python
process = subprocess.Popen(
    ["python"],
    stdin=subprocess.PIPE,
    stdout=subprocess.PIPE,
    text=True
)

output, _ = process.communicate("print(5 + 5)")
print(output)
```

```python
hi = subprocess.run(["dir"], capture_output=True, text=True, shell=True)
print(hi.stdout)
```

```python
subprocess.run(["dirs"], capture_output=True, shell=True, check=True)
```

```python
print(subprocess.check_output(["dir"], shell=True, text=True))
```
