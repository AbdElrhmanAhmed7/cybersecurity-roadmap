```python
import random
import os
from pathlib import Path
import shelve
import time
```

```python
base_dir = Path.home() / "MyStudyHub"
print(f"File dir: {base_dir.absolute()}")
base_dir.mkdir(parents=True, exist_ok=True)
print(
    f"Anchor: {base_dir.anchor}, Drive: {base_dir.drive}, Parts: {base_dir.parts}")
```

```python
notes = base_dir / "Notes"
quizzes = base_dir / "Quizzes"
progress = base_dir / "Progress"
[folder.mkdir(exist_ok=True) for folder in (notes, quizzes, progress)]
(base_dir / "Notes").relative_to(Path.home())
```

```python
with open((notes / "tasks.txt"), "w", encoding="UTF-8") as f:
    print(f.write(
        "Today's Tasks:\n1. Go to the Gym.\n2. Study My lessons.\n3. Help my mom.\n"))

(quizzes / "subjects.txt").write_text("1. History.\n2. Arabic.\n3. English.\n4. Math.\n5. Science.\n")
```

```python
with open(list(notes.glob("tasks.txt"))[0], "a", encoding="UTF-8") as f:
    f.write("4. Go talk to my girl.\n")

with open(list(notes.glob("tasks.txt"))[0], "r", encoding="UTF-8") as f:
    print(f.readlines())

list(quizzes.glob("subjects.txt"))[0].read_text()
```

```python
for i, x in enumerate(base_dir.glob("*")):
    print(f"{i}. {x}")

print("-------")

for i, x in enumerate(base_dir.rglob("*.txt")):
    print(f"{i}. {x}")

print("-------")

for i, x in enumerate(base_dir.glob("Notes/*")):
    print(f"{i}. {x}")
```

```python
with shelve.open(base_dir / "my_settings") as my_shelf:
    my_shelf["subjects"] = list(base_dir.glob(
        "Quizzes/subjects.txt"))[0].read_text()
    my_shelf["streak"] = 5

with shelve.open(base_dir / "my_settings") as my_shelf:
    print(list(my_shelf.keys()))
    print(list(my_shelf.values()))
```
