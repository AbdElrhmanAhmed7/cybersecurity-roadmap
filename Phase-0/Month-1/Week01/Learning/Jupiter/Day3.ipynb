```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)


factorial(int(input("n: ")))
```

```python
def fibonacci(n):
    if n == 1 or n == 0:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)


fibonacci(int(input("n: ")))
```

```python
def is_palindrome(s):
    while True:
        if len(s) == 1:
            return True
        elif s[0] == s[-1]:
            s = s[1:-1]
        else:
            return False


is_palindrome(input("Enter: "))
```
