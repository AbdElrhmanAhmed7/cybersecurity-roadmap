```python
with open("../Python/exampleWithHeader3.csv", "r") as f:
    reader = csv.reader(f)
    for line in reader:
        line = list(line)
        for col in range(len(line)):
            print(f"Row #{reader.line_num}: Column #{col + 1}: {line[col]}")
```

```python
with open("exampleWithHeader3.csv", "a", newline="") as f:
    writer = csv.writer(f)
    print(writer.writerow(['Hello, world!', 'eggs', 'bacon', 1234]))
```

```python
with open("exampleWithHeader3.csv", "r") as f:
    reader = csv.DictReader(f)
    for line in reader:
        print(line)
```

## Exercise for (Chapter 12 Automate the boring stuff).


```python
from pathlib import Path
```

```python
def find_csv_files(directory):
    yield Path(directory).absolute().rglob("*.csv")


def read_csv_content_without_header(csv_directory):
    with open(csv_directory, "r") as f:
        content = csv.reader(f)
        next(content)
        return list(content)


def write_csv_content(csv_directory):
    with open(csv_directory, "w") as f:
        csv_writer = csv.writer(f)
        csv_writer.writerows(read_csv_content_without_header(csv_directory))
    print(f"Removed header from {csv_directory.name}.")
```

```python
for csv_file in find_csv_files("."):
    write_csv_content(csv_file)
```
